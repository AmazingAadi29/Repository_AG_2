Use matplotlib library to plot a sine function ~~~~~~~

````python
import matplotlib.pyplot as plt
x = np.arange(0, 2*np.pi, 0.01) # np,arange(min, max, increments)
y = np.cos(x)
z = np.sin(x)

# plotting two curves on the same graph
plt.plot(x, y)
plt.plot(x, z)

plt.show()
````

````python
#subplot()

fig, axs = plt.subplots(2, 1)

axs[0].plot(x, y)
#axs[0].set_title('Sine')

axs[1].plot(x, z)
#axs[1].set_title('Cosine')

#plt.tight_layout()
plt.show()
````

#Assignment 1 :P

````python
# grades for 5 students in 3 classes
grades = np.array([
[85, 90, 78], # student 1
[92, 88, 84], # student 2
[76, 85, 91], # student 3
[88, 77, 95], # student 4
[94, 89, 82] # student 5
])
````

Task 1: Calculate Statistics - avg grade for each student and subject, calculate and print average of the whole class


````python
# subject averages - prints the average for each subject
subject1_avg = np.mean(grades, 0, int)
print('subject averages:', subject1_avg)

# student averages - prints the GPA for each student
students_avg = np.mean(grades, axis = 1)
print('students GPA:', students_avg)

# school average - prints the aveage grade students in the school
school_avg = np.mean(grades)
print('school average:', school_avg)
````

Task 2: Apply Broadcasting - students get extra credit for each subject as a bonus. Use broadcasting for this and explain how it is used

````python
# broadcasting is essentially a way for NumPy to do math with different size arrays.
# in this case, we will be using a small 1x3 array and adding it to a 5x3 array
bonus = np.array([2, 4, 7])
bonus.shape # array size of the extra credit
grades.shape # array size of the grade, comparing it to the bonus

print(bonus + grades) # the 1x3 array is added to the 5x3 array, adding "2, 4, and 7" to each respective column
````

Task 3: Plot Data - Create a bar plot of the average grades vs. students using Matplotlib and label the axes and title

````python
import matplotlib.pyplot as plt
# creating an array of students to plot each average grade with
students = ['Aadi', 'Youfus', 'Abyaan', 'Sid', 'Saxon']
# students = x-axis, students_avg = y-axis
plt.bar(students, students_avg)

plt.xlabel('Students') # labels the x-axis 'students'
plt.ylabel('Grades') # labels the y-axis 'grades'
plt.title('Students and their grades') # adds a title 'Students and their grades'

plt.show()
````
